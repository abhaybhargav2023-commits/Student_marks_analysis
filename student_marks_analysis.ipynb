# Student Marks Analysis

## Objective
Analyze student marks using Python lists
and calculate the total, average, highest,
and lowest marks.

## Concepts
- Python lists
- sum(), len(), max(), min()
- Basic statistical calculations

```python
marks = [88, 92, 79, 92, 85]

print("students marks : ", marks)
print("Number of students :", len(marks))
```

```python
total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)

print("Total marks :", total)
print("Average marks :", average)
print("Highest marks :", highest)
print("Lowest marks :", lowest)
```
